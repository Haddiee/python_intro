```python
# CLASS ACTIVITY 2.1b 
# Write a program that checks if a number is positive, negative, or zero using proper indentation. Use if, elif, and else.

number = 1  # Try changing this to 0 or a positive number

if number > 0:
    print(number, "is positive")
elif number < 0:
    print(number, "is negative")
else:
    print("Zero")
```

```python
# CODE EXAMPLE 2.3a: All the ways to use print()

student_name = "Larundu"
student_age = 1.9
gpa = 5.0

# 1. Print string literals
print("====Student Details====")

# 2. Print a variable alone
print(student_name)

# 3. Print multiple items (comma-separated)
print("Name:", student_name)
print("Age:", student_age)
print("GPA:", gpa)

# 4. Print an empty line as a separator
print()

# 5. Print a number directly (no quotes needed for numbers)
print("In five years", student_name, "will be", student_age + 5)
```
